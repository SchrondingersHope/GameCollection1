```python
attributes = df.columns.tolist()[:-1]
target_attribute = df.columns[-1]
print(attributes)
df.columns
```

```python
#calcultating entropy
from math import log2

def entropy(df1, target):
  total = len(df1)

  #the following method seems to be the fastest
  positive_samples_p = ((df1[target] == "Yes").sum())/ total
  negative_samples_p = 1 - positive_samples_p

  if positive_samples_p == 1 or negative_samples_p == 1:
    return 0

  entr = -(negative_samples_p)*log2(negative_samples_p) - (positive_samples_p)*log2(positive_samples_p)

  return entr

entropy(df, target_attribute)
entropy(df[df['Outlook']=='Sunny'],target_attribute)


#all the code that didn't make it, idk im attached

#negative_samples_p = (len(df1) - positive_samples_p) / total
#df[target_attribute].value_counts()  alt way to calculate num of rows for each value of target_attribute
#print('\n', type(positive_samples_p),'\n',type(target_attribute))
```

```python
#calculating info gain
def info_gain(df2, attr, targ2):
  unique_values = df2[attr].unique()
  counts = df2[attr].value_counts()

  gain = entropy(df2,targ2)

  total = len(df2)

  for x in unique_values:
    #print(x, '\n', type(x))
    #print(counts[x], '\n', x)
    gain -= (entropy(df2[df2[attr] == x],targ2) * counts[x]/total)

  return gain


for att in attributes:
  print(info_gain(df, att, target_attribute))

info_gain(df, 'Outlook', target_attribute)
info_gain(df, 'Temperature',target_attribute)
```

```python
class Node:
  def __init__(self, feature=None, label=None):
    self.feature = feature
    self.label = label
    self.children = {}
    self.edges = []

  def is_leaf(self) -> bool:
    return self.label is not None

root = Node(feature= attributes[0])
root.children = df['Outlook'].unique()
print(root.children)

test_row = df.iloc[1]
test_row[attributes[0]]

#len(df[target_attribute].unique())
```

```python
#Time for id3

def id3(df3, attributes3, target_attribute3):

  if( len( df3[target_attribute3].unique() ) == 1):
    return Node(label=df3[target_attribute3].iloc[0])

  if( len(attributes3)==0 ):
    most_common_value = df3[target_attribute3].mode()[0]
    return Node(label=most_common_value)

  info_gain_vals = [info_gain(df3, attr3, target_attribute3) for attr3 in attributes3]
  besti = info_gain_vals.index(max(info_gain_vals))
  best_attr = attributes3[besti]

  root = Node(feature=best_attr)
  root.edges = df3[best_attr].unique()

  for edg in root.edges:
    examples = df3[df3[best_attr]==edg]

    if examples.empty:
      child = Node(label=df3[target_attribute3].mode()[0])
    else:
      rest_attributes = [att for att in attributes3 if att!=best_attr]

      child = id3(examples,rest_attributes,target_attribute3)

    root.children[edg] = child

  return root


tree = id3(df, attributes, target_attribute)
```

Let's check the tree using pprint first

```python
from pprint import pprint

def tree_to_dict(node):
    if node.is_leaf():
        return node.label
    else:
        return {
            node.feature: {
                edg: tree_to_dict(child) for edg, child in node.children.items()
            }
        }

tree_dict = tree_to_dict(tree)
pprint(tree_dict, sort_dicts=False)
```

Works! Time to use the tree to classify the examples and check for classification accuracy

```python
def classify(df4_row, node4):
  row = df4_row

  if(node4.is_leaf()):
    return node4.label

  edg = row[node4.feature]
  return classify(row, node4.children[edg])

true_pred = 0
for i in range(len(df)):
  pred = classify(df.iloc[i], tree)
  if(pred == df.iloc[i][target_attribute]):
    true_pred +=1

  print(pred," ", df.iloc[i][target_attribute])

print("Accuracy of the DT is :", true_pred/len(df))

```

```python
import numpy as np

n = 100
actual = np.random.binomial(1, 0.5,n)
pred = np.random.binomial(1,0.5, n)

true_p = false_p = true_n = false_n = 0
for i in range(n):
  if(actual[i]):
    if(pred[i]):
      true_p += 1
    else:
      false_n += 1
  elif(pred[i]):
    false_p +=1
  else:
    true_n +=1

confusion_mat = np.array([[true_p , false_n],[false_p, true_n]])
accuracy = (true_p + true_n ) /n
error = 1- accuracy
precision = true_p / (true_p + false_p)
sensitivity = true_p / (true_p + false_n)
specificity = true_n / (true_n + false_p)
f_score = 2 * precision * sensitivity / (sensitivity +  precision)


print(actual , '\n' , pred)
print(confusion_mat)

print({"Accuracy": accuracy, "Error": error, "Precision": precision, "Sensitivity": sensitivity, "Specificity": specificity, "F-Score": f_score})
```
